```python
def clean_col_names(df):
    """
    Updates column names to remove spaces and make lower case
    INPUT: complete dataframe
    OUTPUT: complete dataframe with new col names
    """
    df2 = df.copy()
    df2.columns = df2.columns.map(lambda x: x.lower().replace(' ','_'))
    return df2

def rename_cols(df):
    df2 = df.copy()
    df2 = df2.rename(columns={"total_time_spent_on_website": "time_on_website", 
                        "how_did_you_hear_about_x_education": "heard_about_x_ed",
               "what_is_your_current_occupation":"current_occupation",
               "what_matters_most_to_you_in_choosing_a_course":"most_imp_factor","receive_more_updates_about_our_courses":"get_updates",
               "i_agree_to_pay_the_amount_through_cheque": "pay_by_cheque" ,"a_free_copy_of_mastering_the_interview":"interview_ebook",
              "update_me_on_supply_chain_content":"supply_chain_updates"})
    return df2
```

```python
data = clean_col_names(data)
data.columns
data = rename_cols(data)
```

```python
data.drop(columns=["asymmetrique_activity_index","asymmetrique_profile_index","asymmetrique_activity_score","asymmetrique_profile_score"])
```
